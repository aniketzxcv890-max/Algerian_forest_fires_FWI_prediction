```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df=pd.read_csv('Algerian_forest_fires_dataset.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df[df.isnull().any(axis=1)]
```

the dataset is converted into teo sets based on region from 122th , we can make a new column based on the region
1. "bejaia region datset"
2. "sidi-bel abbes region dataset"
Add new column with region

```python
df.loc[:122,'region']=0
df.loc[122:,'region']=1
```

```python
df.head()
```

```python
df.info()
```

```python
df[['region']]= df[['region']].astype(int)
```

```python
df = df.dropna().reset_index(drop = True)
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
df = df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns = df.columns.str.strip()  # to remove spaces 
```

```python
df.columns
```

```python
df[['month','year','day','Temperature','RH','Ws']] = df[['month','year','day','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

```python
objects = [features for features in df.columns if df[features].dtypes =='O']
```

```python
for i in objects:
    if i!='Classes':
        df[i]=df[i].astype(float)
```

```python
df.info()
```

```python
df.head()
```

```python
## Let save the cleaned dataset
df.to_csv('Algerian_forest_fires_cleaned_dataset.csv',index=False)
```


 Exploratory Data Analysis

```python
## drop day,month and year
df_copy=df.drop(['day','month','year'],axis=1)
```

```python
df_copy.head()
```

```python
## categories in classes
df_copy['Classes'].value_counts()
```

```python
## Encoding of the categories in classes
df_copy['Classes']=np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df_copy.tail()
```

```python
## Plot desnity plot for all features
plt.style.use('seaborn-v0_8')
df_copy.hist(bins=50,figsize=(20,15))
plt.show()
```

```python
## Percentage for Pie Chart
percentage=df_copy['Classes'].value_counts(normalize=True)*100
```

```python
# plotting piechart
classlabels=["Fire","Not Fire"]
plt.figure(figsize=(12,7))
plt.pie(percentage,labels=classlabels,autopct='%1.1f%%')
plt.title("Pie Chart of Classes")
plt.show()
```

```python
## Box Plots
sns.boxplot(df['FWI'],color='green')
plt.show()
```

```python
df['Classes']=np.where(df['Classes'].str.contains('not fire'),'not fire','fire')
```

```python
## Monthly Fire Analysis
dftemp=df.loc[df['region']==1]
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',hue='Classes',data=df)
plt.ylabel('Number of Fires',weight='bold')
plt.xlabel('Months',weight='bold')
plt.title("Fire Analysis of Sidi- Bel Regions",weight='bold')
plt.show()
```

```python
## Monthly Fire Analysis
dftemp=df.loc[df['region']==0]
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',hue='Classes',data=dftemp)
plt.ylabel('Number of Fires',weight='bold')
plt.xlabel('Months',weight='bold')
plt.title("Fire Analysis of Brjaia Regions",weight='bold')
plt.show()
```
